```python
%sql
SELECT 
    p.PRODUCT_NUMBER, 
    pl.PRODUCT_LINE_EN AS PRODUCT_LINE, 
    pt.PRODUCT_TYPE_EN AS PRODUCT_TYPE, 
    pnl.PRODUCT_NAME AS PRODUCT_NAME, 
    SUM(od.QUANTITY) AS TOTAL_QUANTITY,
    SUM(od.QUANTITY * od.UNIT_PRICE) AS REVENUE
FROM 
    tutorials.gosales.order_details od
JOIN 
    tutorials.gosales.product p ON od.PRODUCT_NUMBER = p.PRODUCT_NUMBER
JOIN 
    tutorials.gosales.product_type pt ON p.PRODUCT_TYPE_CODE = pt.PRODUCT_TYPE_CODE
JOIN 
    tutorials.gosales.product_line pl ON pt.PRODUCT_LINE_CODE = pl.PRODUCT_LINE_CODE
JOIN 
    tutorials.gosales.product_name_lookup pnl ON p.PRODUCT_NUMBER = pnl.PRODUCT_NUMBER
GROUP BY 
    p.PRODUCT_NUMBER, pl.PRODUCT_LINE_EN, pt.PRODUCT_TYPE_EN, pnl.PRODUCT_NAME
ORDER BY 
    TOTAL_QUANTITY DESC
LIMIT 3;
```

```python
%sql
WITH RankedProducts AS (
    SELECT 
        p.PRODUCT_NUMBER, 
        pl.PRODUCT_LINE_EN AS PRODUCT_LINE, 
        pt.PRODUCT_TYPE_EN AS PRODUCT_TYPE, 
        pnl.PRODUCT_NAME AS PRODUCT_NAME, 
        SUM(od.QUANTITY) AS TOTAL_QUANTITY,
        SUM(od.QUANTITY * od.UNIT_PRICE) AS REVENUE,
        ROW_NUMBER() OVER (PARTITION BY pl.PRODUCT_LINE_EN ORDER BY SUM(od.QUANTITY * od.UNIT_PRICE) DESC) AS rank
    FROM 
        tutorials.gosales.order_details od
    JOIN 
        tutorials.gosales.product p ON od.PRODUCT_NUMBER = p.PRODUCT_NUMBER
    JOIN 
        tutorials.gosales.product_type pt ON p.PRODUCT_TYPE_CODE = pt.PRODUCT_TYPE_CODE
    JOIN 
        tutorials.gosales.product_line pl ON pt.PRODUCT_LINE_CODE = pl.PRODUCT_LINE_CODE
    JOIN 
        tutorials.gosales.product_name_lookup pnl ON p.PRODUCT_NUMBER = pnl.PRODUCT_NUMBER
    GROUP BY 
        p.PRODUCT_NUMBER, pl.PRODUCT_LINE_EN, pt.PRODUCT_TYPE_EN, pnl.PRODUCT_NAME
)
SELECT 
    PRODUCT_NUMBER, 
    PRODUCT_LINE, 
    PRODUCT_TYPE, 
    PRODUCT_NAME, 
    TOTAL_QUANTITY, 
    REVENUE
FROM 
    RankedProducts
WHERE 
    rank <= 3
ORDER BY 
    PRODUCT_LINE, rank;
```

```python
%sql
WITH RankedProducts AS (
    SELECT 
        p.PRODUCT_NUMBER, 
        pl.PRODUCT_LINE_EN AS PRODUCT_LINE, 
        pt.PRODUCT_TYPE_EN AS PRODUCT_TYPE, 
        pnl.PRODUCT_NAME AS PRODUCT_NAME, 
        SUM(od.QUANTITY) AS TOTAL_QUANTITY,
        SUM(od.QUANTITY * od.UNIT_PRICE) AS REVENUE,
        ROW_NUMBER() OVER (PARTITION BY pl.PRODUCT_LINE_EN ORDER BY SUM(od.QUANTITY * od.UNIT_PRICE) DESC) AS rank
    FROM 
        tutorials.gosales.order_details od
    JOIN 
        tutorials.gosales.product p ON od.PRODUCT_NUMBER = p.PRODUCT_NUMBER
    JOIN 
        tutorials.gosales.product_type pt ON p.PRODUCT_TYPE_CODE = pt.PRODUCT_TYPE_CODE
    JOIN 
        tutorials.gosales.product_line pl ON pt.PRODUCT_LINE_CODE = pl.PRODUCT_LINE_CODE
    JOIN 
        tutorials.gosales.product_name_lookup pnl ON p.PRODUCT_NUMBER = pnl.PRODUCT_NUMBER AND pnl.PRODUCT_LANGUAGE = 'EN'
    GROUP BY 
        p.PRODUCT_NUMBER, pl.PRODUCT_LINE_EN, pt.PRODUCT_TYPE_EN, pnl.PRODUCT_NAME
)
SELECT 
    PRODUCT_NUMBER, 
    PRODUCT_LINE, 
    PRODUCT_TYPE, 
    PRODUCT_NAME, 
    TOTAL_QUANTITY, 
    REVENUE
FROM 
    RankedProducts
WHERE 
    rank <= 3
ORDER BY 
    PRODUCT_LINE, rank;
```
